# Linear Regression:
### y = mx + c
```
Here,
    m = slope
        It measures the change in the mean value of y
        If, 
            m = (+)ve -> y increases as x increases
            m = (-)ve -> y decreases as x increases
        
    c = The y-intercept.
        It is the mean value of y when x equals 0

    
x	y	  xy	  x^2
6	350	  2100	  36
8	775	  6200	  64
12	1150  13800	  144
14	1395  19530	  196
18	1675  30150	  324


Here,

avg_x = (6+8+12+14+18) / 5 = 11.6
avg_y = (350+775+1150+1395+1675) / 5 = 1069
avg_xy = (2100+6200+13800+19530+30150) / 5 = 14356
(avg_x)^2 = 134.56
avg_x^2 = (36+64+144+196+324) / 5 = 152.8

m = ((avg_x*avg_y) - avg_xy) / ((avg_x)^2 - avg_x^2)
c = avg_y - (m*avg_x)


y = mx + c

```
    

````python
df['xy'] = df['x']*df['y']
df
````

````python
df['x^2'] = df['x']**2
df
````

````python
avg_x = df['x'].mean()
avg_y = df['y'].mean()
avg_x_sq = df['x^2'].mean()
avg_xy = df['xy'].mean()
````

````python
m = (((avg_x*avg_y) - avg_xy) / (avg_x**2 - avg_x_sq)).round(2)
m
````

````python
c = (avg_y - (m*avg_x)).round(2)
c
````

````python
x = 17
y = (m*x + c).round(2)

y
````
