```python
import numpy as np
import seaborn as sns
```

```python
arr = np.array([2,4,7,10,15,20,25,205])
```

```python
np.mean(arr)
```

```python
np.median(arr)
```

```python
sns.boxplot(x = arr)
```

```python
n = len(arr)

mn = arr[0]
mx = arr[n-1]

if n%2 != 0:
    Q2 = arr[n//2]
else:
    Q2 = (arr[n//2 -1]+arr[n//2]) / 2
    
lower = arr[:n//2]
upper = arr[n//2:] if n%2==0 else arr[n//2 +1:]

m = len(lower)

if m%2 != 0:
    Q1 = lower[m//2]
else:
    Q1 = (lower[m//2 -1]+lower[m//2])/2

m = len(upper)

if m%2 != 0:
    Q3 = upper[m//2]
else:
    Q3 = (upper[m//2 -1]+upper[m//2])/2

print(f"Minimum: {mn}\nQ1: {Q1}\nQ2: {Q2}\nQ3: {Q3}\nMaximmum: {mx}")
```

```python
IQR = Q3-Q1

upper_fence = Q3 + 1.5 * IQR
lower_fence = Q1 - 1.5 * IQR

print(lower_fence, upper_fence)
```

```python
arr = np.array([i for i in arr if (i>=lower_fence) and (i<=upper_fence)])

print(*arr)
```

```python
sns.boxplot(x=arr)
```
