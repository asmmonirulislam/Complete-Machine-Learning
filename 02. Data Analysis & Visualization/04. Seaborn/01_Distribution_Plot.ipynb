# histplot
### What it shows:
* The frequency distribution of a numerical variable using bars
### Main Application
* How the values are distributed?
### Python
* sns.histplot(data, bins=int, kde=True)

```python
plt.subplot(1, 2, 1)
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.subplot(1, 2, 2)
sns.histplot(df['tip'], bins=20, kde=True)
```

# kdeplot

```python
sns.kdeplot(df['total_bill'])
```

# jointplot
### What it shows:
* The relationship between two numerical variables, along with their individual distributions.
### Usually includes:
* Central scatter plot
* Distribution of X
* Distribution of Y
### Main Application:
* How are two numerical variables related, and how is each individually distributed?
### Python
* sns.jointplot(data=df, x='total_bill', y='tip', kind='scatter', hue='category')

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='scatter', hue='sex')
```

# pairplot
### What is shows:
* Pairwise relationships among multiple numerical variables in a dataset
### Main Applications
* How does every numerical feature relate to every other numerical feature?
### pairplot helps identify:
* Strong relationships
* Possible Linear Patterns
* Clusters
* Outliers
* Seaprability between categories
### Python
* sns.pairplot(df)

```python
sns.pairplot(df, hue='sex')
```
