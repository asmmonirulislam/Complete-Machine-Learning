```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Categorical Data Plots
There are a few main plot types for this:
* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

```python
df = sns.load_dataset('tips')
```

```python
df
```

# countplot
### What it shows:
* The number of observation in each category
### Main Application
* How many records belong to each category?
### Python
sns.countplot(data=df, x=column_name)

```python
df['sex'].value_counts()
```

```python
df['smoker'].value_counts()
```

```python
pd.crosstab(df['sex'], df['smoker'])
```

```python
sns.countplot(x = df['smoker'], hue=df['sex'])
```

# barplot
### What it shows:
* The aggregated numerical  value of a variable for each category
### Main Application:
* How does the avaerage of a numerical variable differ across categories?
### Python
* sns.barplot(x = categorical_datatype, y = quantitative_datatype, estimator=mean/median/sum/std)

```python
sns.barplot(x = df['sex'], y = df['tip'])   # default: np.mean
```

```python
sns.barplot(x = df['sex'], y = df['tip'], estimator=np.sum)     # based on np.sum --> sum of tip
```

# boxplot
### What it shows:
Compares the quantitative data with the categorical data using:
* Median
* Q1
* Q3
* IQR
* Whiskers
* Potential Outliers

```python
sns.boxplot(x = df['tip'], y = df['day'])
```

# violinplot
### Compares the quantitative data with the categorical data

```python
sns.violinplot(x = df['day'], y = df['tip'])
```

# stripplot

```python
sns.stripplot(x = df['day'], y = df['total_bill'])
```

# swarmplot

```python
sns.swarmplot(x = df['day'], y = df['total_bill'])
```

```python
sns.violinplot(x = df['day'], y = df['tip'])
sns.swarmplot(x = df['day'], y = df['tip'])
```
