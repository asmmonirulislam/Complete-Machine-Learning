```python
import seaborn as sns
import numpy as np
import pandas as pd
import cufflinks as cf
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips
```

# px.line(data_frame=None, x=None, y=None)

```python
px.line(data_frame=tips, y='total_bill')
```

```python
px.bar(data_frame=tips, x='tip', y='total_bill')
```

```python
px.scatter(data_frame=tips, x='tip', y='total_bill')
```

```python
grp_tip = tips.groupby('day')['tip'].mean()
grp_tip
```

```python
px.bar(grp_tip)
```

```python
grp_smoker = tips.groupby(['sex', 'smoker']).size()
grp_smoker
```

```python
px.bar(tips.groupby('sex')['smoker'].size())
```

```python
px.bar(tips['smoker'].value_counts())
```
