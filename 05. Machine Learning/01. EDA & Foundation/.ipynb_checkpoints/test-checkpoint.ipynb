```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.isnull().sum()*100
```

```python
df=df.dropna()
```

```python
df.shape
```

```python
non_num_cols
```

```python
df['smoker'].unique()
```

```python
df['smoker'] = df['smoker'].map({'yes':1, 'no':0})
df.head()
```

```python
df['sex'].unique()
```

```python
df['sex'] = df['sex'].map({'female':0, 'male':1})
df.head()
```

```python
df['region'].unique()
```

```python
df = pd.get_dummies(df, columns=['region'])
df.head()
```

```python
region_cols = [col for col in df.columns if col.startswith('region_')]
df[region_cols] = df[region_cols].astype(int)
df.head()
```

```python
df.info()
```

```python
df = df.rename(columns={
    'sex':'is_male',
    'smoker':'is_smoker'
})
df.columns
```

```python
df.head()
```

```python
labels = ['underweight', 'normal', 'overweight', 'obesity_1', 'obesity_2', 'severe_obesity']
bins = [0.0, 18.5, 25.0, 30.0, 35.0, 40.0, float('inf')]
df['bmi_category'] = pd.cut(
    df['bmi'],
    bins=bins,
    labels=labels
)
df.head()
```

```python
df.groupby('bmi_category')['bmi'].agg(['min', 'max'])
```

```python
df.groupby('bmi_category')['bmi'].agg(
    range = lambda x: x.max()-x.min()
)
```

```python
df.info()
```

```python
df = pd.get_dummies(df, columns=['bmi_category'])
```

```python
df
```

```python
bmi_cols = [col for col in df.columns if col.startswith('bmi_category_')]
df[bmi_cols] = df[bmi_cols].astype(int)
df.head()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(df.corr(numeric_only=True), annot=True)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler_cols = ['age', 'bmi', 'children']
df[scaler_cols] = scaler.fit_transform(df[scaler_cols])
df.head()
```

```python
from scipy.stats import pearsonr

cols = [col for col in df.columns if col != 'charges']
correlation = {}
for col in cols:
    correlation[col]= pearsonr(df[col], df['charges'])[0]
```

```python
df_corr = pd.DataFrame(correlation.items(), columns=['Features', 'Correlation']).sort_values(by='Correlation', ascending=False).reset_index(drop=True)
```

```python
df_corr
```

```python
plt.figure(figsize=(35,15))
plt.yticks(np.arange(-2, 2, 0.1))
sns.barplot(data=df_corr, x='Features', y='Correlation')
```
