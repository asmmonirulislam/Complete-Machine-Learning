# Conditional probability 
### is the probability of an event occurring given that another event has already occurred.
### It measures the likelihood of event A occurring when event B is known to have occurred.
### Formula:
### P(A|B) = P(A∩B) / P(B)

* P(A∣B) = Probability of event A given event B.
* P(A∩B) = Probability of both A and B occurring (intersection).
* P(B) = Probability of event B, where P(B)>0.

## Example:
### Suppose a student has a 60% probability of passing Mathematics and a 40% probability of passing both Mathematics and Physics.
### P(Math | Physics) = P(Math ∩ Physics) / P(Physics) = 40 / 60 = 0.6666666666666666
