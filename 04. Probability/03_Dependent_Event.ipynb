# Dependent Event:
### Occurence of one event affects the occurence of other

## Mathematically:
### P(A ∩ B) = P(A) * P(B|A)

## Example
### Drawing cards without placement
* Suppose you have a deck of 52 cards
* Let A be the event of drawing an ACE on the first draw
* Let B be the event of drawing another ACE on the second draw without putting the first card back

```python
total_possibilities = 52
fav_outcome = 4
pA = fav_outcome/total_possibilities
total_possibilities -= 1
fav_outcome -= 1
pBA = fav_outcome / (total_possibilities)
prob = pA * pBA
prob
```
