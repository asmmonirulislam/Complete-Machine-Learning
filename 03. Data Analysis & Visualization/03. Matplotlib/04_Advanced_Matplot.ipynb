```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.linspace(1, 5, 11)
b = a**2
c = a*2
d = c+10
```

MATPLOT BASIC

```python
plt.plot(a, b)
plt.title('Title of the plot')
plt.xlabel('a')
plt.ylabel('b')
```

SUBPLOT IN MATPLOT

plt.subplots(row_size, col_size, positional_value)

```python
plt.subplot(2,2,1) # 1st position
plt.plot(a, b)
plt.subplot(2,2,2) # 2nd position
plt.plot(b,c)
plt.subplot(2,2,3) # 3rd position
plt.plot(c,d)
plt.subplot(2,2,4) # 4th position
plt.plot(d,a)
plt.show()
```

```python
plt.subplot(1,3,1)
plt.subplot(1,3,2)
plt.subplot(1,3,3)
```

OBJECT ORIENTED WAY

```python
# plt.figure(num=None, figsize=(float, float), dpi=float, facecolor=background_color, edgecolor=border_color, frameon=True)
# plt.figure().add_axes(rect,projection=None, polar=False, **kwargs)
# rect -> tuple(left, bottom, width, height) : [The dimensions of the axes]

fig = plt.figure(figsize=(6,6), dpi=100)

axes1 = fig.add_axes((0.7, 0.7, 0.5, 0.5))
axes1.plot(a, b)

axes2 = fig.add_axes((0.1, 0.7, 0.5, 0.5))
axes2.plot(c, d)

axes3 = fig.add_axes((0.1, 0.1, 0.5, 0.5))
axes3.plot(c, d)

axes4 = fig.add_axes((0.7, 0.1, 0.5, 0.5))
axes4.plot(c, d)
```

TYPES OF PLOT

```python
# Scatter

plt.scatter(a, b)
```

```python
# hist

import random
data = random.sample(range(1, 1000), 100)
plt.hist(data)
```

```python
# boxplot

plt.boxplot(a)
```

SAVE PLOT

```python
fig = plt.figure()
ax = fig.add_axes((0.1, 0.1, 0.8, 0.8))
ax.plot(a, b)
plt.savefig('save-plot')
```
