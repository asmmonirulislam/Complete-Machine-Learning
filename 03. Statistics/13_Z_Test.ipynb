### We use Z-Test when the sample size is more than 30

## Formula
```
Z = (x̄ - μ) / (σ / √n)

x̄ = Sample mean
μ = Population mean
σ = Population standard deviation
n = Sample Count

```

````python
sample_mean = 74
population_mean = 70
population_standard_deviation = 10
sample_size = 36
````

````python
import numpy as np


z_test = (sample_mean - population_mean) / (population_standard_deviation / np.sqrt(sample_size))
z_test
````

# Z Test Code

````python
import numpy as np
from scipy.stats import norm  #norm = normal distributions
````

````python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
sample_count = len(sample)
````

````python
z_score = (sample_mean-population_mean) / (population_std/np.sqrt(sample_count))
z_score
````

````python
p_value = 2 * (1-norm.cdf(abs(z_score)))
p_value
````

````python
alpha = 0.05
if p_value<alpha:
    print(f"I will reject the null hypothesis")
else:
    print(f"I will accept the null hypothesis")
````
