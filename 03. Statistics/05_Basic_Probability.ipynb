# Probability
### Probability is the measure of the likelihood or chance that an event will occur.
## Formula
### P_E = Number of favourable outcomes / Total number of possible outcomes


#

### When tossing a fair coin, the probability of getting heads is:

```python
fav_outcomes = 1
total_outcomes = 2
p_e = fav_outcomes / total_outcomes
p_e
```

### It is expressed as a number between 0 and 1, where:
* 0 represents an impossible event and
* 1 represents a certain event.
