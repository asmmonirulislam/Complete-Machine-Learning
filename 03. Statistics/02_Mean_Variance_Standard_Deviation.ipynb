```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
arr = np.array([85, 90, 95, 100, 105])
```

```python
arr
```

```python
N = np.size(arr)

Mean = (np.sum(arr)/N)

var_total = 0
for item in arr:
    var_total += (Mean-item)**2
Variance = var_total/N
```

```python
Standard_Deviation = np.sqrt(Variance)
```

```python
print(f"Mean: {Mean}\nVariance: {Variance}\nStandard Deviation: {Standard_Deviation.round(2)}")
```
