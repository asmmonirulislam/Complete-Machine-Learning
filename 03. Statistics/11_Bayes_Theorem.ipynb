# Bayes Theorem
#### Bayes theorem is a formula used to calculate the probability of an event based on new evidence or information


### Formula

### P(A|B) = ( P(B|A) * P(A) ) / P(B) ; where

* P(A|B) = Probability of A given B → posterior probability
* P(B|A) = Probability of B given A → likelihood
* P(A) = Probability of A before seeing evidence → prior probability
* P(B) = Probability of the evidence B → evidence

### Example

###  Student and Exam

#### Suppose a class has 100 students.

* 60 students studied for the exam.
* 40 students did not study.
* Among the students who studied, 50 passed.
* Among the students who did not study, 10 passed.

#### Now, we randomly select a student and find out that the student passed.

### Question

* What is the probability that the student studied?

```python
studied = 60
not_studied = 40
passed_studied = 50
passed_not_studied = 10

total = studied + not_studied

P_studied = studied / total

print(f"P(studied): {P_studied}")

P_passed_given_studied = passed_studied / studied 

print(f"P(passed | studied): {P_passed_given_studied}")

P_passed = (passed_studied + passed_not_studied) / total

print(f"P(passed): {P_passed}")

P_studied_given_passed = (P_passed_given_studied * P_passed) / P_studied

print(f"P(studied | passed): {round(P_studied_given_passed*100, 2)}%")
```
